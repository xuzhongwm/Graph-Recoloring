## NetworkX Tutorial For Recoloring

```python
import networkx as nx

G = nx.Graph()
G.add_edge("A", "B")
G.add_edge("B", "C")
G.add_edge("C", "A")
G.add_edge("C", "D")
G.add_node(1)
```

We are able to visualize G with matplot 

```python
import matplotlib.pyplot as plt

nx.draw(G, font_weight='bold')
```
